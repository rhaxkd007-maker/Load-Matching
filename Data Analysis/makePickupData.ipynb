```python
delivery_id = df_pickup['delivery_user_id'].unique().tolist()
```

```python
# 이름과 성의 조합으로 전체 이름을 만듬
all_comb = [f"{ln} {fn}" for ln, fn in product(last_names, first_names)]

#랜덤하게 선택되게 하기위해
random.seed(42)
random.shuffle(all_comb)
# delivery_id 갯수만큼만 필요
names = (all_comb * (len(delivery_id) // len(all_comb) + 1))[:10484]
names
```
